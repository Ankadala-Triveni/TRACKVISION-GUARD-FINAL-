# RailGuard AI â€” Notebook 03: Comprehensive Model Evaluation

This notebook evaluates the performance of the trained **YOLO11 Nano (`yolo11n.pt`) model** across both the **Validation** and **Test** dataset splits.

### Objectives:
1. Load evaluation metrics summary (`evaluation_summary.json`).
2. Compare overall Precision ($P$), Recall ($R$), $mAP_{50}$, and $mAP_{50-95}$ across splits.
3. Inspect full 15-class per-class detection metric breakdown table.
4. Analyze Confusion Matrix heatmaps (`confusion_matrix.png`).
5. Evaluate rare class performance (`corrosion_low`, `broken_rail_medium`).

```python
import os
import json
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

# Define paths
RUN_DIR = os.path.abspath("../ai-service/runs/yolo11_railguard_v1")
SUMMARY_JSON = os.path.join(RUN_DIR, "evaluation_summary.json")
CONF_MATRIX_IMG = os.path.join(RUN_DIR, "confusion_matrix.png")

print(f"Run Directory: {RUN_DIR}")
print(f"Summary JSON Exists: {os.path.exists(SUMMARY_JSON)}")
print(f"Confusion Matrix Image Exists: {os.path.exists(CONF_MATRIX_IMG)}")
```

## 1. Overall Dual-Split Evaluation Metrics (Validation vs Test)

```python
if os.path.exists(SUMMARY_JSON):
    with open(SUMMARY_JSON, "r") as f:
        summary_data = json.load(f)
        
    overall_metrics = summary_data.get("overall", {})
    df_overall = pd.DataFrame(overall_metrics)
    print("=== Overall Model Metrics ===")
    print(df_overall.to_string(index=False))
else:
    print("Evaluation summary JSON will be generated automatically once training finishes.")
```

## 2. Full 15-Class Per-Class Metric Breakdown

```python
if os.path.exists(SUMMARY_JSON):
    per_class_val = summary_data.get("per_class_val", [])
    per_class_test = summary_data.get("per_class_test", [])
    
    df_val = pd.DataFrame(per_class_val)
    df_test = pd.DataFrame(per_class_test)
    
    print("=== Validation Split â€” Per-Class Metrics ===")
    print(df_val.to_string(index=False))
    
    print("\n=== Test Split â€” Per-Class Metrics ===")
    print(df_test.to_string(index=False))
```

## 3. Confusion Matrix Heatmap

```python
if os.path.exists(CONF_MATRIX_IMG):
    img = Image.open(CONF_MATRIX_IMG)
    plt.figure(figsize=(12, 10))
    plt.imshow(img)
    plt.title("RailGuard AI â€” Confusion Matrix Heatmap", fontsize=14, fontweight="bold")
    plt.axis("off")
    plt.show()
else:
    print("Confusion Matrix plot will be rendered upon completion of model evaluation.")
```
